```python
import random
import pandas as pd
import numpy as np
from tqdm import tqdm
from faker import Faker
from sklearn.utils import shuffle

fake = Faker()
```

```python
products = {
    "Mobile": [
        "Prepaid Basic 5GB", "Prepaid Plus 20GB", "Prepaid Unlimited",
        "Postpaid Silver 50GB", "Postpaid Gold 100GB", "Postpaid Platinum"
    ],
    "Internet": [
        "ZENDFiber Home 100 Mbps", "ZENDFiber Home 300 Mbps", "ZENDFiber Home 1 Gbps",
        "ZENDOffice Net 200", "ZENDOffice Net 500", "ZENDOffice Net 1G"
    ],
    "Business": [
        "ZENDBiz Connect 100", "ZENDBiz Connect 500", "ZENDBiz Connect 1G",
        "ZENDEnterprise Ultra", "ZENDEnterprise Dedicated"
    ],
    "Cloud": [
        "ZENDCloud VM Basic", "ZENDCloud VM Pro", "ZENDCloud VM Enterprise",
        "ZENDStorage 1TB", "ZENDStorage 10TB", "ZENDArchive Storage"
    ],
    "IoT": [
        "ZENDSmart Traffic", "ZENDSmart Lighting", "ZENDSmart Parking",
        "ZENDIndustrial Sensor", "ZENDFleet IoT"
    ]
}

countries = ["India", "USA", "Singapore", "Thailand"]
user_types = ["individual", "enterprise"]

policies = {
    "billing": "ZENDS follows monthly billing in advance and suspends services after 7 days of late payment.",
    "refund": "Full refund is allowed within 7 days if usage is less than 10 percent. Cloud services are non-refundable after activation.",
    "sla": "Individual users have 98.5 percent uptime SLA, business users 99.5 percent, and enterprise users 99.9 percent.",
    "contract": "Enterprise customers have a minimum 12-month contract period.",
    "fair_usage": "Unlimited plans are capped at 1TB per month under fair usage policy."
}

intents = ["Billing", "Refund", "Technical", "Complaint", "Product Inquiry"]
sentiments = ["Angry", "Neutral", "Happy"]
```

```python
def build_text(intent):

    category = random.choice(list(products.keys()))
    product = random.choice(products[category])
    country = random.choice(countries)
    user = random.choice(user_types)
    sentiment = random.choice(sentiments)

    intro = random.choice([
        f"I am a {user} customer from {country}.",
        f"This is regarding my {product} service.",
        f"I recently subscribed to {product} in {country}.",
        f"My account is registered as a {user} user in {country}."
    ])

    if intent == "Billing":
        core = random.choice([
            f"I noticed unexpected charges for {product}.",
            f"My invoice amount seems incorrect.",
            f"There is a discrepancy in my latest bill.",
            f"I was charged even after payment was made."
        ])
        policy_ref = policies["billing"]

    elif intent == "Refund":
        core = random.choice([
            f"I want to request a refund for {product}.",
            f"I cancelled my subscription but did not receive refund.",
            f"Please initiate refund process immediately.",
            f"I am eligible under refund policy but refund not processed."
        ])
        policy_ref = policies["refund"]

    elif intent == "Technical":
        core = random.choice([
            f"The service is not working properly.",
            f"I am facing connectivity issues.",
            f"Speed is extremely slow.",
            f"There is complete service outage."
        ])
        policy_ref = policies["sla"]

    elif intent == "Complaint":
        core = random.choice([
            f"I am extremely dissatisfied with the service.",
            f"Customer support has not responded.",
            f"The issue remains unresolved for weeks.",
            f"This service quality is unacceptable."
        ])
        policy_ref = policies["contract"]

    else:  # Product Inquiry
        core = random.choice([
            f"I would like to know features and pricing details.",
            f"Does this plan include SLA guarantee?",
            f"Please explain benefits compared to other plans.",
            f"Is this plan suitable for enterprise use?"
        ])
        policy_ref = policies["fair_usage"]

    text = f"{intro} {core} As per policy, {policy_ref}"

    if sentiment == "Angry":
        text = "This is unacceptable! " + text + " Please resolve this urgently."
    elif sentiment == "Happy":
        text = "Thank you for your support. " + text

    return text, intent, sentiment
```

```python
def generate_dataset(n=20000):
    data = set()

    while len(data) < n:
        intent = random.choice(intents)
        record = build_text(intent)
        data.add(record)

    df = pd.DataFrame(list(data), columns=["text", "intent", "sentiment"])
    df = shuffle(df).reset_index(drop=True)
    return df

df = generate_dataset(20000)

print("Final dataset size:", len(df))
```
